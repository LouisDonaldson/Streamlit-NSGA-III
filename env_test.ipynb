```python
from pymoo.algorithms.moo.mopso_cd import MOPSO_CD
from pymoo.problems import get_problem
from pymoo.optimize import minimize
from pymoo.visualization.scatter import Scatter

problem = get_problem("zdt1")

algorithm = MOPSO_CD(
    pop_size=100,
    w=0.6,
    c1=2.0,
    c2=2.0,
    max_velocity_rate=0.5,
    archive_size=200,
    seed=1
)

res = minimize(problem,
               algorithm,
               ('n_gen', 200),
               seed=1,
               verbose=False)

plot = Scatter()
plot.add(problem.pareto_front(), plot_type="line", color="black", alpha=0.7)
plot.add(res.F, facecolor="none", edgecolor="red")
plot.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

# Data
algorithms = ["NSGA-II", "NSGA-III", "MOE/D", "SMS-EMOA", "MOPSO-CD"]
config1 = [12.57, 12.15, 12.20, 12.42, 12.08]
config2 = [43.32, 44.45, 45.66, 44.02, 22.67]
config3 = [108.57, 107.94, 55.66, 104.05, 84.87]

x = np.arange(len(algorithms))
width = 0.35

fig, ax = plt.subplots(figsize=(10, 5))
ax.bar(x - width/2, config1, width, label="Config 1 (TTC in s)")
ax.bar(x + width/2, config2, width, label="Config 2 (TTC in s)")
ax.bar(x + width, config3, width, label="Config 3 (TTC in s)")

ax.set_ylabel("TTC (s)")
ax.set_title("MOOP Algorithm Comparison (across 7 days)")
ax.set_xticks(x)
ax.set_xticklabels(algorithms)
ax.legend()

plt.tight_layout()
plt.show()
```

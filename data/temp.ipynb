```python
import pandas as pd

import pandas as pd

def build_daily_averages(input_csv, output_csv):
    # Load CSV with header
    df = pd.read_csv(input_csv)

    # Convert the Date column to datetime
    df["Date"] = pd.to_datetime(df["Date"], errors="coerce")

    # Extract the actual calendar date (YYYY-MM-DD)
    df["actual_day"] = df["Date"].dt.date

    # Assign sequential day numbers based on when the date changes
    df["day_number"] = (df["actual_day"] != df["actual_day"].shift()).cumsum()

    # Compute averages for each day (numeric columns only)
    daily_avg = df.groupby("day_number").mean(numeric_only=True)

    # Save to CSV
    daily_avg.to_csv(output_csv)
    daily_avg.to_json("daily_averages.json", orient="records", indent=2)

    print(f"Saved daily averages to {output_csv}")


build_daily_averages("wave_data.csv", "daily_averages.csv")
```
